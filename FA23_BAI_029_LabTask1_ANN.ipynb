### Task 1: Variables and Basic Operators

This task demonstrates variable creation, printing data types, using arithmetic operators, comparison operators, and swapping variables.

```python
name, roll_no, cgpa = "Alice", 101, 3.8
print(name, type(name))
print(roll_no, type(roll_no))
print(cgpa, type(cgpa))

marks1, marks2 = 80, 90
marks1 += 5
marks2 -= 3
print(marks1, marks2)

print(marks1 == marks2)
print(marks1 > marks2)

print("Before:", marks1, marks2)
marks1, marks2 = marks2, marks1
print("After:", marks1, marks2)
```

### Task 7: Loops with Break and Continue

This task covers loop control syntax, including skip actions using continue, exit commands using break, while loops, and range aggregations.

```python
for i in range(1, 31):
    if i % 4 == 0: continue
    print(i, end=" ")
print()

for i in range(1, 51):
    if i > 20: break
    print(i, end=" ")
print()

n = 5
i = 1
while i <= 10:
    print(f"{n} x {i} = {n*i}")
    i += 1

print(sum(i for i in range(1, 101) if i % 2 == 0))
```

### Task 9: Classes and Inheritance

This task illustrates object-oriented programming, defining classes, calling methods, and inheriting and extending parent classes using super().

```python
class Course:
    def __init__(self, course_name, credit_hours):
        self.course_name = course_name
        self.credit_hours = credit_hours
    def display_info(self):
        print(self.course_name, self.credit_hours)

c1 = Course("AIC380", 3)
c2 = Course("CS101", 4)
c1.display_info()
c2.display_info()

class LabCourse(Course):
    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours
    def display_info(self):
        super().display_info()
        print(self.lab_hours)

lc = LabCourse("PHY102", 4, 3)
lc.display_info()
```
