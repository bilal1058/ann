```python
# Lab Task 1: Building, training, and experimenting with a modified CNN on MNIST

import tensorflow as tf
from tensorflow.keras import models, layers, Input
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# 1. Load the MNIST dataset
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

# 2. Preprocess the image data: reshape to include the channel dimension (grayscale = 1) and normalize pixels to [0, 1]
train_images = train_images.reshape((60000, 28, 28, 1)).astype('float32') / 255
test_images = test_images.reshape((10000, 28, 28, 1)).astype('float32') / 255

# 3. One-hot encode the target labels (10 classes for digits 0-9)
train_labels = to_categorical(train_labels)
test_labels = to_categorical(test_labels)

# 4. Design the CNN Model with alternative architecture choices:
# - Using an explicit Input layer to avoid modern Keras warnings
# - Experimenting with 3 Convolutional layers with increasing filter sizes (32 -> 64 -> 128)
# - Adding Dropout to prevent overfitting
model_experiment = models.Sequential([
    Input(shape=(28, 28, 1)),

    # First Convolutional Block
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Second Convolutional Block
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Third Convolutional Block (Experimenting with additional depth)
    layers.Conv2D(128, (3, 3), activation='relu'),

    # Flatten the 3D output to 1D before feeding into dense layers
    layers.Flatten(),

    # Dense Hidden Layer with Dropout Regularization
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.5),

    # Output classification layer with softmax activation for 10 classes
    layers.Dense(10, activation='softmax')
])

# Display the model summary
model_experiment.summary()

# 5. Compile the model using the 'adam' optimizer as an experiment (faster convergence than 'sgd')
model_experiment.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# 6. Train the experimental model
batch_size = 128
epochs = 5 # Using 5 epochs for quick and efficient evaluation of design choices

history = model_experiment.fit(
    train_images, train_labels,
    batch_size=batch_size,
    epochs=epochs,
    validation_split=0.1,  # Using 10% of training data for validation during training
    verbose=1
)

# 7. Evaluate the model performance on the unseen test dataset
test_loss, test_acc = model_experiment.evaluate(test_images, test_labels, verbose=0)
print(f"\nExperiment Test Accuracy: {test_acc:.4f}")
print(f"Experiment Test Loss: {test_loss:.4f}")
```

```python
import tensorflow as tf
from tensorflow.keras import models, layers, Input, optimizers
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# --- Load and Preprocess Dataset to prevent NameError ---
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
train_images = train_images.reshape((60000, 28, 28, 1)).astype('float32') / 255
test_images = test_images.reshape((10000, 28, 28, 1)).astype('float32') / 255
train_labels = to_categorical(train_labels)
test_labels = to_categorical(test_labels)

# --- Lab Task 2: Enhancing Neural Network Performance ---
# We will build an enhanced CNN architecture and use an advanced optimizer (RMSprop with momentum).
# We will also tune hyperparameters such as units, batch size, and epoch count.

# 1. Define the enhanced architecture
model_enhanced = models.Sequential([
    Input(shape=(28, 28, 1)),

    # First Convolutional block with 32 filters
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    # Second Convolutional block with increased capacity (64 filters)
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),

    # Densely connected classifier with tuned layer size (128 neurons)
    layers.Dense(128, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.4), # Regularization to prevent overfitting

    # Output layer for 10-class classification
    layers.Dense(10, activation='softmax')
])

# 2. Compile using an Advanced Optimizer: RMSprop with custom learning rate and momentum
opt = optimizers.RMSprop(learning_rate=0.001, momentum=0.9)
model_enhanced.compile(
    optimizer=opt,
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Display the model architecture
model_enhanced.summary()

# 3. Train with tuned hyperparameters: Batch Size of 64 and 8 Epochs
tuned_batch_size = 64
tuned_epochs = 8

history_enhanced = model_enhanced.fit(
    train_images, train_labels,
    batch_size=tuned_batch_size,
    epochs=tuned_epochs,
    validation_split=0.15,
    verbose=1
)

# 4. Evaluate the optimized model's performance on the test set
test_loss_enhanced, test_acc_enhanced = model_enhanced.evaluate(test_images, test_labels, verbose=0)
print(f"\nEnhanced Model Test Accuracy: {test_acc_enhanced:.4f}")
print(f"Enhanced Model Test Loss: {test_loss_enhanced:.4f}")
```
